```python
# ==============================================================================
# ANÁLISIS UNIFICADO: COSTO ABSOLUTO VS. COSTO NORMALIZADO (Z-SCORE)
# ==============================================================================
# Objetivo: Generar el gráfico de distribución original y una segunda versión
# con los costos normalizados para comparar la forma de las distribuciones.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore # Importamos la función para calcular Z-Score

# Suponiendo que 'df' es tu DataFrame ya cargado y con la limpieza inicial.

if 'df' in locals() and not df.empty:
    print("Iniciando la generación de ambos gráficos de distribución unificada...")

    # ==========================================================================
    # FIGURA 1: DISTRIBUCIÓN COMPARATIVA CON COSTOS ABSOLUTOS (El que ya tenemos)
    # ==========================================================================
    plt.figure(figsize=(14, 8))
    sns.set_theme(style="whitegrid")

    sns.histplot(data=df, x="FIRST_ALT_VALUE_sum", element="poly", fill=False, color='#4A70B0', linewidth=2.5, label='Alternativa 1')
    sns.histplot(data=df, x="SECOND_ALT_VALUE_sum", element="poly", fill=False, color='#9B9B9B', linewidth=2.5, label='Alternativa 2')
    sns.histplot(data=df['THIRD_ALT_VALUE_sum'].dropna(), element="poly", fill=False, color='#E84855', linewidth=2.5, label='Alternativa 3')

    plt.title('Figura 1: Distribución Unificada de Costos Absolutos', fontsize=18, weight='bold', pad=20)
    plt.xlabel('Costo Total de Peajes ($)', fontsize=14, labelpad=15)
    plt.ylabel('Frecuencia de Rutas', fontsize=14, labelpad=15)
    plt.legend(title='Alternativa', fontsize=12, title_fontsize=13)
    current_values = plt.gca().get_xticks()
    plt.gca().set_xticklabels([f'{int(x/1000)}k' for x in current_values])
    plt.tight_layout()
    plt.show()

    # ==========================================================================
    # FIGURA 2: DISTRIBUCIÓN COMPARATIVA CON COSTOS NORMALIZADOS (Z-SCORE)
    # ==========================================================================
    
# --- 1. Calcular los Z-Scores manualmente para preservar la estructura ---
# Este es el método más robusto en pandas.

# --- Para Alternativa 1 ---
mean1 = df['FIRST_ALT_VALUE_sum'].mean() # mean() ignora NaNs por defecto
std1 = df['FIRST_ALT_VALUE_sum'].std()   # std() ignora NaNs por defecto
df['Z_SCORE_ALT1'] = (df['FIRST_ALT_VALUE_sum'] - mean1) / std1

# --- Para Alternativa 2 ---
mean2 = df['SECOND_ALT_VALUE_sum'].mean()
std2 = df['SECOND_ALT_VALUE_sum'].std()
df['Z_SCORE_ALT2'] = (df['SECOND_ALT_VALUE_sum'] - mean2) / std2

# --- Para Alternativa 3 ---
mean3 = df['THIRD_ALT_VALUE_sum'].mean()
std3 = df['THIRD_ALT_VALUE_sum'].std()
df['Z_SCORE_ALT3'] = (df['THIRD_ALT_VALUE_sum'] - mean3) / std3

print("Cálculo manual de Z-Scores completado correctamente.")

# --- 2. Visualizar las nuevas distribuciones normalizadas ---
# El código de visualización no necesita cambios.
plt.figure(figsize=(14, 8))

sns.histplot(data=df, x="Z_SCORE_ALT1", element="poly", fill=False, color='#4A70B0', linewidth=2.5, label='Alternativa 1')
sns.histplot(data=df, x="Z_SCORE_ALT2", element="poly", fill=False, color='#9B9B9B', linewidth=2.5, label='Alternativa 2')
sns.histplot(data=df, x="Z_SCORE_ALT3", element="poly", fill=False, color='#E84855', linewidth=2.5, label='Alternativa 3')

plt.title('Figura 2: Distribución Unificada de Costos Normalizados (Z-Score)', fontsize=18, weight='bold', pad=20)
plt.xlabel('Costo Normalizado (Desviaciones Estándar desde la Media)', fontsize=14, labelpad=15)
plt.ylabel('Frecuencia de Rutas', fontsize=14, labelpad=15)
plt.legend(title='Alternativa', fontsize=12, title_fontsize=13)
plt.axvline(0, color='black', linestyle='--', linewidth=1) 
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# ANÁLISIS DE ANOMALÍAS (VERSIÓN CON R-CUADRADO)
# ==============================================================================
# Objetivo: Enriquecer el gráfico de regresión con la métrica R-cuadrado
# para cuantificar la fuerza de la relación.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score # Importamos la función para calcular R-cuadrado

if 'df' in locals() and not df.empty:
    print("\nIniciando análisis de anomalías (con R-cuadrado)...")

    # --- 1. Preparar y Filtrar los datos (como antes) ---
    if 'Costo_Viaticos' not in df.columns:
        df['Costo_Viaticos'] = df['FOOD_VALUE_first'].fillna(0) + df['LODGING_VALUE_first'].fillna(0)

    df_corregido = df.copy()
    df_corregido['Costo_Viaticos'] = df_corregido['Costo_Viaticos'].replace(0, np.nan)
    
    df_analisis_final = df_corregido[['DURATION_HRS_first', 'Costo_Viaticos']].dropna()
    print(f"Se analizarán {df_analisis_final.shape[0]} viajes con datos completos y válidos.")

    # --- 2. Crear y Entrenar el Modelo (como antes) ---
    X = df_analisis_final[['DURATION_HRS_first']]
    y = df_analisis_final['Costo_Viaticos']

    model = LinearRegression()
    model.fit(X, y)

    # --- 3. Calcular el R-cuadrado ---
    # Hacemos las predicciones sobre los mismos datos con los que entrenamos el modelo
    y_pred = model.predict(X)
    
    # Calculamos el valor de R-cuadrado
    r_squared = r2_score(y, y_pred)
    print(f"\nEl valor de R-cuadrado (R²) es: {r_squared:.2f}")

    # --- 4. Visualizar la relación ---
    plt.figure(figsize=(12, 8))
    ax = sns.regplot(x='DURATION_HRS_first', y='Costo_Viaticos', data=df_analisis_final,
                     scatter_kws={'alpha':0.6, 'color':'#4A70B0'}, line_kws={'color':'#E84855', 'linestyle':'--'})
    
    plt.title('Duración vs. Viáticos', fontsize=16, pad=20)
    plt.xlabel('Duración del Viaje (Horas)', fontsize=12)
    plt.ylabel('Costo de Viáticos', fontsize=12)
    plt.grid(True)

    # --- AJUSTE CLAVE: AÑADIR EL TEXTO DEL R-CUADRADO ---
    # Creamos el texto y lo posicionamos en una esquina del gráfico.
    # 'ax.transAxes' nos permite usar coordenadas relativas (0,0 es abajo a la izquierda, 1,1 es arriba a la derecha).
    plt.text(0.05, 0.95, f'$R^2 = {r_squared:.2f}$',
             transform=ax.transAxes,
             fontsize=14,
             verticalalignment='top',
             bbox=dict(boxstyle='round,pad=0.5', fc='wheat', alpha=0.5))

    # --- AJUSTE CLAVE: AÑADIR LA NOTA EXPLICATIVA ---
    # 1. Definimos el texto. Usamos '\n' para crear un salto de línea y que no sea muy largo.
    nota_texto = ("Nota: El área sombreada representa el Intervalo de Confianza del 95% para la regresión,\n"
                  "indicando el rango de incertidumbre de la tendencia estimada.")

    # 2. Añadimos el texto al gráfico en la esquina inferior izquierda.
    plt.text(0.35, 0.02, nota_texto,
             transform=ax.transAxes, # Usamos coordenadas relativas
             fontsize=10,
             verticalalignment='bottom', # Alineamos el texto desde su base
             bbox=dict(boxstyle='round,pad=0.3', fc='lightgray', alpha=0.6)) # Añadimos un fondo sutil
    plt.show()

    # (El resto del análisis de anomalías seguiría aquí como antes)

else:
    print("El DataFrame 'df' está vacío o no existe.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import FuncFormatter

# --- 1. CONFIGURACIÓN INICIAL ---
INPUT_FILE_NAME = 'top_frequent_routes_adt.xlsx'
OUTPUT_FILE_NAME = 'export_analisis_kilometraje.xlsx'

# --- 2. CARGA Y PREPARACIÓN DE DATOS ---
try:
    df = pd.read_excel(INPUT_FILE_NAME)
    print(f"Archivo '{INPUT_FILE_NAME}' cargado exitosamente.\n")

    column_rename_map = {
        'KM MAPS': 'KM_MAPS',
        'KM WAZE': 'KM_WAZE',
        'km Via Michelin': 'KM_VIA_MICHELIN'
    }
    df.rename(columns=column_rename_map, inplace=True)
    
    cols_for_avg = ['KM_MAPS', 'KM_WAZE', 'KM_VIA_MICHELIN']
    df['KM_PROMEDIO_REF'] = df[cols_for_avg].mean(axis=1)
    print("Datos preparados y columna de promedio calculada.\n")
    
    
    
    
    # --- 3. CÁLCULOS DE ANÁLISIS ---
    primary_source_columns = ['N_KM_MODE', 'KM_MAPS', 'KM_WAZE', 'KM_VIA_MICHELIN']
    min_global = df[primary_source_columns].min().min()
    max_global = df[primary_source_columns].max().max()
    for col in primary_source_columns:
        norm_col_name = f"{col}_norm"
        df[norm_col_name] = (df[col] - min_global) / (max_global - min_global)

    base_column = 'N_KM_MODE'
    reference_columns = ['KM_MAPS', 'KM_WAZE', 'KM_VIA_MICHELIN', 'KM_PROMEDIO_REF']
    for col in reference_columns:
        ref_name = col.replace('KM_', '').replace('PROMEDIO_', 'PROM_')
        df[f"DIF_KM_{ref_name}"] = df[col] - df[base_column]
        df[f"DIF_%_{ref_name}"] = ((df[col] / df[base_column]) - 1) * 100
        
    cols_to_round = [col for col in df.columns if df[col].dtype == 'float64']
    df[cols_to_round] = df[cols_to_round].round(3)
    print("Análisis numérico completado (Normalización y Desviaciones).\n")
    
    
    
    
    
    # --- 4. CÁLCULO DEL IMPACTO MONETARIO (COP) ---
    # Se verifica si la columna de costo existe antes de realizar el cálculo
    if 'TOT_PAID_RUTA_KEY_ANNUALIZED' in df.columns:
        print(f"Calculando impacto monetario (COP) basado en la columna '{'TOT_PAID_RUTA_KEY_ANNUALIZED'}'.\n")
        
        # Lista de sufijos para iterar sobre las columnas de referencia
        reference_suffixes = ['MAPS', 'WAZE', 'VIA_MICHELIN', 'PROM_REF']
        
        for suffix in reference_suffixes:
            # Nombre de la columna de diferencia porcentual
            perc_diff_col = f"DIF_%_{suffix}"
            # Nombre de la nueva columna para el impacto monetario (formato profesional)
            money_impact_col = f"IMPACTO_COP_{suffix}"
            
            # Fórmula: Aplicar el porcentaje de diferencia al costo total
            df[money_impact_col] = (df[perc_diff_col] / 100) * df['TOT_PAID_RUTA_KEY_ANNUALIZED']
            
        print("Cálculo de impacto monetario completado.\n")
    else:
        # Advertencia si la columna de costo no se encuentra
        print(f"ADVERTENCIA: La columna de costo '{'TOT_PAID_RUTA_KEY_ANNUALIZED'}' no fue encontrada.")
        print("El cálculo del impacto monetario será omitido.\n")
        
    
    
        
    # --- 5. CÁLCULO Y REPORTE DE AHORROS POTENCIALES ---
    # Esta nueva sección calcula la suma total de los impactos negativos (ahorros) por cada fuente.
    
    print("--- Resumen de Ahorros Potenciales Totales (Sumando Impactos Negativos) ---\n")
    
    # Se usan los mismos sufijos que en el paso anterior para iterar
    impact_suffixes_for_sum = ['MAPS', 'WAZE', 'VIA_MICHELIN', 'PROM_REF']
    
    # Se verifica que las columnas de impacto existan antes de intentar sumar
    if f'IMPACTO_COP_{impact_suffixes_for_sum[0]}' in df.columns:
        for suffix in impact_suffixes_for_sum:
            impact_col_name = f'IMPACTO_COP_{suffix}'       
            
            # 1. Filtrar solo los valores negativos (ahorros) y sumarlos
            total_ahorro = df[df[impact_col_name] < 0][impact_col_name].sum()
            
            # 2. Crear el identificador solicitado para el reporte
            id_reporte = f"TOT_IMPACTO_{suffix}"
            
            # 3. Imprimir el resultado con formato de moneda (separador de miles y 2 decimales)
            print(f"{id_reporte}: {total_ahorro:,.2f} COP")

        print("\n----------------------------------------------------------------------\n")
    
    else:
        print("No se calcularon los ahorros potenciales porque las columnas de impacto monetario no existen.\n")



        

    # --- 5. LIMPIEZA FINAL Y EXPORTACIÓN ---
    # Redondear todas las columnas de tipo float a 3 decimales para un reporte limpio
    cols_to_round = [col for col in df.columns if pd.api.types.is_float_dtype(df[col])]
    df[cols_to_round] = df[cols_to_round].round(3)
    print("Redondeo final aplicado a todas las columnas numéricas.\n")
    
    

    # --- 4. EXPORTACIÓN DEL ARCHIVO EXCEL ---
    # df.to_excel(OUTPUT_FILE_NAME, index=False)
    # print(f"El archivo con los resultados numéricos ha sido guardado como: '{OUTPUT_FILE_NAME}'")
  





    # --- 5. VISUALIZACIÓN ---
    print("Iniciando Fase de Visualización...\n")
    sns.set_theme(style="whitegrid")

    # Paso 5.1: Gráfico de Boxplots Comparativos
    print("Iniciando Fase de Visualización: Dashboard de Boxplots...\n")
    sns.set_theme(style="whitegrid")

    # Paso 5.1: Dashboard de Boxplots Comparativos (Unificados)
    # Creamos una única figura con dos subgráficos (1 fila, 2 columnas)
    fig, axes = plt.subplots(1, 2, figsize=(22, 9))
    fig.suptitle('Análisis Comparativo de Distribuciones de Kilometraje', fontsize=20, fontweight='bold')

    # Gráfico Izquierdo: Boxplots con Valores Absolutos (Reales)
    abs_columns_to_plot = ['N_KM_MODE', 'KM_MAPS', 'KM_WAZE', 'KM_VIA_MICHELIN']
    df_melted_abs = df.melt(value_vars=abs_columns_to_plot, var_name='Fuente de Datos', value_name='Kilómetros')
    
    sns.boxplot(ax=axes[0], x='Fuente de Datos', y='Kilómetros', data=df_melted_abs, color="seagreen")
    axes[0].set_title('Análisis de Magnitud Absoluta (Valores Reales)', fontsize=16, fontweight='bold')
    axes[0].set_xlabel('Fuente de Datos de Kilometraje', fontsize=12)
    axes[0].set_ylabel('Kilómetros', fontsize=12)
    axes[0].tick_params(axis='x', rotation=0)

    # Gráfico Derecho: Boxplots con Valores Normalizados
    norm_columns_to_plot = [f"{col}_norm" for col in primary_source_columns]
    df_melted_norm = df.melt(value_vars=norm_columns_to_plot, var_name='Fuente de Datos', value_name='Valor Normalizado')
    
    sns.boxplot(ax=axes[1], x='Fuente de Datos', y='Valor Normalizado', data=df_melted_norm, color="#347AB8")
    axes[1].set_title('Análisis de Consistencia Relativa (Valores Normalizados)', fontsize=16, fontweight='bold')
    axes[1].set_xlabel('Fuente de Datos de Kilometraje', fontsize=12)
    axes[1].set_ylabel('Valores Normalizados (Escala 0 a 1)', fontsize=12)
    axes[1].tick_params(axis='x', rotation=0)

    # Ajustamos el layout y mostramos la figura que contiene ambos gráficos.
    plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Ajuste para que el título principal no se solape
    plt.show()
    
    print("Dashboard de Boxplots Unificados generado exitosamente.")








    # ==============================================================================
    # SECCIÓN A AJUSTAR 1: DEFINICIÓN DE LA FUNCIÓN DE FORMATO
    # ==============================================================================
    # Es una buena práctica definir la función de formato una sola vez aquí arriba.
    # Esta función corrige el problema de los "0k" repetidos.
    def formato_k(x, pos):
        """Formatea los números del eje X. Muestra 'k' para miles."""
        if x >= 1000:
            return f'{int(x/1000)}k'
        elif x == 0:
            return '0' # Para asegurar que el cero se muestre como '0' y no '0k'
        return int(x)
    # ==============================================================================


    # Paso 5.2: Dashboard de Polígonos de Frecuencia
    print("Generando Gráfico 2: Dashboard de Polígonos de Frecuencia...")
    fig, axes = plt.subplots(1, 2, figsize=(24, 8))
    fig.suptitle('Análisis Comparativo de Frecuencias: Visión Absoluta vs. Porcentual', fontsize=18, fontweight='bold')
    
    # ==============================================================================
    # SECCIÓN A AJUSTAR 2: CÓDIGO DEL GRÁFICO IZQUIERDO
    # ==============================================================================
    # Aquí aplicamos las dos correcciones que discutimos.
    
    # Gráfico Izquierdo
    cols_abs = ['N_KM_MODE', 'KM_MAPS', 'KM_WAZE', 'KM_VIA_MICHELIN']
    df_abs_melted = df.melt(value_vars=cols_abs, var_name='Fuente', value_name='Kilómetros')
    
    # AJUSTE 1 (Opcional pero recomendado): Añadir 'common_norm=False'
    # Esto escala cada línea para que el área bajo ella sea 1. Facilita la
    # comparación de las FORMAS de las distribuciones, que es lo que nos interesa.
    sns.histplot(data=df_abs_melted, x='Kilómetros', hue='Fuente', element="poly", fill=False, ax=axes[0], common_norm=False)
    
    axes[0].set_title('Distribución de Kilómetros Absolutos', fontsize=14, fontweight='bold')
    axes[0].set_xlabel('Kilómetros por Ruta', fontsize=12)
    
    # AJUSTE 2: Cambiar la etiqueta del eje Y si usas la normalización
    # Si dejas 'common_norm=False', la etiqueta correcta es "Densidad". Si lo quitas, es "Frecuencia".
    axes[0].set_ylabel('Densidad de Rutas', fontsize=12)
    
    # AJUSTE 3: Aplicar la nueva función de formato que creamos
    axes[0].xaxis.set_major_formatter(FuncFormatter(formato_k))
    # ==============================================================================

    # Gráfico Derecho (ESTA SECCIÓN QUEDA IGUAL)
    cols_perc = ['DIF_%_MAPS', 'DIF_%_WAZE', 'DIF_%_VIA_MICHELIN']
    df_perc_melted = df.melt(value_vars=cols_perc, var_name='Fuente de Referencia', value_name='Desviación (%)')
    sns.histplot(data=df_perc_melted, x='Desviación (%)', hue='Fuente de Referencia', element="poly", fill=False, ax=axes[1])
    axes[1].axvline(0, color='black', linestyle='--', linewidth=1.2)
    axes[1].set_title('Distribución de Desviación Porcentual de la Media', fontsize=14, fontweight='bold')
    axes[1].set_xlabel('Desviación de la Media (%) [Negativo = Sobrecosto]', fontsize=12)
    axes[1].set_ylabel('Frecuencia de Rutas', fontsize=12)
    axes[1].xaxis.set_major_formatter(FuncFormatter(lambda x, pos: f'{int(x)}%'))
    
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()
    print("Gráfico 2 generado exitosamente.")







# ==============================================================================
# (BLOQUE MODIFICADO) Paso 5.3: Boxplot de Desviación Absoluta (Vista Completa)
# ==============================================================================
    print("\nGenerando Gráfico 3: Boxplot de Desviación Absoluta (Vista Completa)...")

    # --- Preparación de Datos (la hacemos una vez y la reutilizamos) ---
    dif_km_cols = ['DIF_KM_MAPS', 'DIF_KM_WAZE', 'DIF_KM_VIA_MICHELIN', 'DIF_KM_PROM_REF']
    df_melted_dif_km = df.melt(value_vars=dif_km_cols, 
                            var_name='Fuente de Comparación', 
                            value_name='Diferencia Absoluta (KM)')

    # --- Inicio del Primer Gráfico ---
    plt.figure(figsize=(14, 9)) # Creamos la figura para el primer gráfico

    # Creamos el boxplot completo
    sns.boxplot(x='Fuente de Comparación', y='Diferencia Absoluta (KM)', 
                data=df_melted_dif_km, palette="viridis")

    plt.axhline(0, color='red', linestyle='--', linewidth=1.5, label='Valor de N_KM_MODE (Referencia)')

    # Títulos y etiquetas
    plt.title('Desviación Absoluta', fontsize=18, fontweight='bold')
    plt.xlabel('Fuente de Datos de Referencia', fontsize=12)
    plt.ylabel('Diferencia en Kilómetros', fontsize=12)
    plt.xticks(rotation=0, ha='right')
    plt.legend()
    plt.tight_layout()

    # Mostramos el primer gráfico
    plt.show() 
    print("Gráfico 3 (Vista Completa) generado exitosamente.")


    # ==============================================================================
    # (NUEVO BLOQUE) Paso 5.4: Boxplot de Desviación Absoluta (Vista Acotada)
    # ==============================================================================
    print("\nGenerando Gráfico 4: Boxplot de Desviación Absoluta (Vista Acotada)...")

    # --- Inicio del Segundo Gráfico ---
    plt.figure(figsize=(14, 9)) # Creamos una NUEVA figura para el segundo gráfico

    # Volvemos a dibujar el mismo boxplot
    sns.boxplot(x='Fuente de Comparación', y='Diferencia Absoluta (KM)', 
                data=df_melted_dif_km, palette="viridis")

    plt.axhline(0, color='red', linestyle='--', linewidth=1.5, label='Valor de N_KM_MODE (Referencia)')

    # --- LA MODIFICACIÓN CLAVE: Acotamos el eje Y ---
    plt.ylim(-100, 100)

    # Títulos y etiquetas para el segundo gráfico
    plt.title('Desviación Absoluta (Rango ±100 KM)', fontsize=18, fontweight='bold')
    plt.xlabel('Fuente de Datos de Referencia', fontsize=12)
    plt.ylabel('Diferencia en Kilómetros', fontsize=12)
    plt.xticks(rotation=0, ha='right')
    plt.legend()
    plt.tight_layout()

    # Mostramos el segundo gráfico
    plt.show()
    print("Gráfico 4 (Vista Acotada) generado exitosamente.")    
    
    
    
    
    
    # ==============================================================================
    # (NUEVO BLOQUE) Paso 5.5: Boxplot de Desviaciones Porcentuales (%)
    # ==============================================================================
    print("\nGenerando Gráfico 5: Boxplot de Desviaciones Porcentuales...")

    # --- Preparación de Datos ---
    # Seleccionamos las columnas de diferencia porcentual
    dif_perc_cols = ['DIF_%_MAPS', 'DIF_%_WAZE', 'DIF_%_VIA_MICHELIN', 'DIF_%_PROM_REF']

    # Derretimos el DataFrame para Seaborn
    df_melted_dif_perc = df.melt(value_vars=dif_perc_cols, 
                                var_name='Fuente de Comparación', 
                                value_name='Diferencia Porcentual (%)')

    # --- Inicio del Gráfico ---
    plt.figure(figsize=(14, 9))

    # Creamos el boxplot. Usamos otra paleta de colores para diferenciar, por ejemplo "coolwarm".
    sns.boxplot(x='Fuente de Comparación', y='Diferencia Porcentual (%)', 
                data=df_melted_dif_perc, palette="coolwarm")

    # La línea de referencia en 0% es crucial
    plt.axhline(0, color='black', linestyle='--', linewidth=1.5, label='Acuerdo Perfecto (0% Desviación)')

    # --- AÑADIMOS FORMATO AL EJE Y para que muestre el símbolo '%' ---
    # Esto hace que el gráfico sea mucho más fácil de leer
    from matplotlib.ticker import FuncFormatter
    plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda y, _: f'{int(y)}%'))

    # Títulos y etiquetas claras
    plt.title('Desviación Porcentual', fontsize=18, fontweight='bold')
    plt.xlabel('Fuente de Datos de Referencia', fontsize=12)
    plt.ylabel('Diferencia Porcentual', fontsize=12)
    plt.xticks(rotation=0, ha='right')
    plt.legend()
    plt.tight_layout()

    # Mostramos el gráfico
    plt.show()
    print("Gráfico 5 (Desviación Porcentual) generado exitosamente.")



    # ==============================================================================
    # (NUEVO BLOQUE) Paso 5.6: Boxplot de Desviaciones Porcentuales (Vista Acotada)
    # ==============================================================================
    print("\nGenerando Gráfico 6: Boxplot de Desviaciones Porcentuales (Vista Acotada)...")

    # --- Inicio del Gráfico Acotado ---
    # No necesitamos preparar los datos de nuevo, ya tenemos la variable 'df_melted_dif_perc'.

    plt.figure(figsize=(14, 9)) # Creamos una NUEVA figura para este gráfico

    # Volvemos a dibujar el mismo boxplot que en el paso anterior
    sns.boxplot(x='Fuente de Comparación', y='Diferencia Porcentual (%)',
                data=df_melted_dif_perc, palette="coolwarm")

    plt.axhline(0, color='black', linestyle='--', linewidth=1.5, label='Acuerdo Perfecto (0% Desviación)')

    # --- LA MODIFICACIÓN CLAVE: Acotamos el eje Y al rango propuesto ---
    plt.ylim(-50, 50)

    # Reaplicamos el formato del eje Y para que se muestre el símbolo '%'
    # Es importante hacerlo de nuevo para este nuevo objeto de gráfico (plt.gca())
    plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda y, _: f'{int(y)}%'))

    # Títulos y etiquetas específicas para el gráfico acotado
    plt.title('Desviación Porcentual (Rango ±50%)', fontsize=18, fontweight='bold')
    plt.xlabel('Fuente de Datos de Referencia', fontsize=12)
    plt.ylabel('Diferencia Porcentual', fontsize=12)
    plt.xticks(rotation=0, ha='right')
    plt.legend()
    plt.tight_layout()

    # Mostramos el nuevo gráfico acotado
    plt.show()
    print("Gráfico 6 (Desviación Porcentual Acotada) generado exitosamente.")





    
# ==============================================================================
# SECCIÓN SIN CAMBIOS
# ==============================================================================
# --- MANEJO DE ERRORES ---
except FileNotFoundError:
    print(f"ERROR: El archivo de entrada '{INPUT_FILE_NAME}' no fue encontrado.")
except KeyError as e:
    print(f"ERROR FATAL: La columna '{e}' es necesaria pero no se encontró.")




# --- MANEJO DE ERRORES ---
# ESTA SECCIÓN QUEDA IGUAL
except FileNotFoundError:
    print(f"ERROR: El archivo de entrada '{INPUT_FILE_NAME}' no fue encontrado.")
except KeyError as e:
    print(f"ERROR FATAL: La columna '{e}' es necesaria pero no se encontró.")
```
